📉Sep 2008 – Mar 2009 (Crisis Financiera)

```python

# Petición de búsqueda y contenido
print("Buscando noticias en Exa... por favor espera.")

resultado = exa.search_and_contents(
    query="Financial analysis of the S&P 500 collapse, Lehman Brothers bankruptcy, and systemic risk in the banking sector.",
    start_published_date="2008-09-01",
    end_published_date="2009-03-31",
    num_results=20,
    include_domains=["reuters.com", "bloomberg.com", "ft.com", "wsj.com",
                     "cnbc.com", "economist.com", "cbs.com", "investing.com","marketwatch.com", "barrons.com", "finance.yahoo.com",
                     "ofdollarsanddata.com", "collabfund.com", "lynalden.com", "klementoninvesting.substack.com", "qoppac.blogspot.com",
                     "awealthofcommonsense.com"],
    text={"include_html_tags": False}  # Texto completo sin HTML ni tablas
)

print("--- ¡BÚSQUEDA COMPLETADA! ---\n")

documentos = []  # Lista donde se acumulan todas las noticias

for noticia in resultado.results:
    fecha_iso = noticia.published_date if noticia.published_date else "Fecha no disponible"

    print(f"TÍTULO: {noticia.title}")
    print(f"ENLACE: {noticia.url}")
    print(f"FECHA DE PUBLICACIÓN: {fecha_iso}")
    print("TEXTO COMPLETO:")
    print(noticia.text if noticia.text else "Texto no disponible")
    print("-" * 50)

    documento = {
        "titulo": noticia.title,
        "url": noticia.url,
        "fecha": fecha_iso,
        "periodo": "caidas crisis financiera",
        "contenido": noticia.text,
        "language": "english",
        "indice_sector": "S&P 500",
        "regimen_mercado":"bearish"
    }
    documentos.append(documento)  # Se añade cada noticia a la lista

# documentos contiene ahora las noticias listas para insertar en la base de datos
print(f"\n{len(documentos)} documentos listos para almacenar.")
```
